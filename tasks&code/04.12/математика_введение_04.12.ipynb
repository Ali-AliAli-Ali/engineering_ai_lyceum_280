# 1. Переменные и функции - как положить слона в холодильник?

```python
animal_species = "слон"     # строка (string)
is_animal_in_fridge = False # булево значение (True/False)
animal_age = 14             # целое число (int)
animal_height = 3.45        # дробное число (float)
animal_name = ["Анна Павловна", "Aнна", "Анечка", "Aннушка"]   # массив (array)
animal_superabilities = ["Стоять на 3 ногах", "Рисовать", "Играть в теннис", "Помещаться в холодильник"]

def pretty_print_ability(ability):
    print("   - ", ability)
pretty_print_function = pretty_print_ability  # функция (callable)!
```

```python
def print_introduction_fridge(species, name, age, height, superabilities, is_in_fridge, format_ability):
    print(f"Встречайте - {species} по имени {name[0]}!")
    print(f"Она отзывается на обращения: {name[1:]}.")

    print(f"Ей {age} лет, она {height} м ростом и умеет много интересного:")
    for ability in superabilities:  # цикл с количеством шагов
        format_ability(ability)

    if not is_in_fridge:            # условие
        print(f"И {name[0]} не в холодильнике)")
    else:
        print(f"И {name[0]} в холодильнике, ого!")
```

```python
print_introduction_fridge(
    animal_species, 
    animal_name, 
    animal_age, 
    animal_height, 
    animal_superabilities, 
    is_animal_in_fridge,
    pretty_print_function
)
```

# 2. Функции - как сделать что угодно?

```python
import random
```

```python
# Пример 1: Функция без входных данных
def random_number_generator():
    """Генерируем случайное число от 1 до 10"""
    return f"Сгенерировал рандомное число: {random.randint(0, 10)}"

# Пример 2: Функция с входными данными
def greeting(name, greeting):
    """Создаем персональное приветствие"""
    return f"{greeting}, {name}! Рад тебя видеть!"

# Пример 3: Функция, "меняющая" входные данные
def double_string(input_str):
    """Заменяем строку на ее же, написанную дважды"""
    input_str += input_str 
    return input_str

# Пример 4: Функция, возвращающая значения из набора
def mood_rating(rate_from_1_to_10):
    """Оцениваем настроение по шкале от 1 до 10"""
    if rate_from_1_to_10 >= 8:
        return "Отличное настроение! 😊"
    elif rate_from_1_to_10 >= 5:
        return "Нормальное настроение 🙂"
    else:
        return "Могло быть лучше... Напишешь, что ты чувствуешь, в своем личном дневнике?"
```

# 3. Графики функций - как увидеть невидимое?

```python
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, widgets
```

### 1. Задаем входные данные...

```python
# Создаем данные для графиков
x = np.linspace(-10, 10, 100)  # 100 точек от -10 до 10
```

### 2. Вычисляем значения разных функций...

```python
linear_func = 2*x + 3
square_func = x**2 - 4
sinus_func = 5*np.sin(x)
exponent_func = np.exp(x/3)
```

### 3. Строим картинки!

```python
# Создаем графики
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Линейная функция
axes[0, 0].plot(x, linear_func, 'b-', linewidth=2)
axes[0, 0].set_title('Линейная: y = 2x + 3')
axes[0, 0].grid(True, alpha=0.3)
axes[0, 0].set_xlabel('x')
axes[0, 0].set_ylabel('y')

# Квадратичная функция
axes[0, 1].plot(x, square_func, 'r-', linewidth=2)
axes[0, 1].set_title('Квадратичная: y = x² - 4')
axes[0, 1].grid(True, alpha=0.3)
axes[0, 1].set_xlabel('x')
axes[0, 1].set_ylabel('y')

# Синус
axes[1, 0].plot(x, sinus_func, 'g-', linewidth=2)
axes[1, 0].set_title('Синусоида: y = 5*sin(x)')
axes[1, 0].grid(True, alpha=0.3)
axes[1, 0].set_xlabel('x')
axes[1, 0].set_ylabel('y')

# Экспонента
axes[1, 1].plot(x, exponent_func, 'm-', linewidth=2)
axes[1, 1].set_title('Экспонента: y = e^(x/3)')
axes[1, 1].grid(True, alpha=0.3)
axes[1, 1].set_xlabel('x')
axes[1, 1].set_ylabel('y')

plt.tight_layout()
plt.show()
```

```python
def interactive_linear(k=1, b=0):
    """Интерактивная функция для построения графика y = kx + b"""
    plt.figure(figsize=(8, 5))
    
    x = np.linspace(-10, 10, 100)
    y = k*x + b
    
    plt.plot(x, y, 'b-', linewidth=3, label=f'y = {k}x + {b}')
    plt.axhline(y=0, color='k', linestyle='-', alpha=0.3)
    plt.axvline(x=0, color='k', linestyle='-', alpha=0.3)
    plt.grid(True, alpha=0.3)
    plt.title(f'График линейной функции: y = {k}x + {b}')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.legend()
    plt.xlim(-10, 10)
    plt.ylim(-10, 10)
    plt.show()
```

```python
interact(
    interactive_linear,
    k=widgets.FloatSlider(value=1, min=-5, max=5, step=0.5, description='k (угол):'),
    b=widgets.FloatSlider(value=0, min=-5, max=5, step=0.5, description='b (сдвиг):')
)
```

# 4. Векторы и матрицы - что такое нейросеть?

```python
import numpy as np
```

```python
print("В машинном обучении данные часто представляют в виде векторов и матриц.")
print("\n🔹 ВЕКТОР - это список чисел (например, оценки ученика)")
print("🔹 МАТРИЦА - это таблица чисел (например, оценки всего класса)")

# Создаем вектор и матрицу
print("\n📊 Пример 1: Создание вектора (списка оценок)")
student_marks = np.array([5, 4, 3, 5, 4])  # Оценки по 5 предметам
print(f"Оценки ученика: {student_marks}")
print(f"Тип: {student_marks.shape} (5 элементов)")
print(f"Средняя оценка: {np.mean(student_marks):.2f}")

print("\n📊 Пример 2: Создание матрицы (оценки всего класса)")
class_marks = np.array([
    [5, 4, 3, 5, 4],  # Ученик 1
    [3, 4, 5, 4, 3],  # Ученик 2
    [4, 5, 4, 4, 5],  # Ученик 3
    [5, 5, 4, 5, 4]   # Ученик 4
])
print(f"Оценки класса:\n{class_marks}")
print(f"Размер матрицы: {class_marks.shape} (4 строки, 5 столбцов)")
print(f"Средняя оценка по классу: {np.mean(class_marks):.2f}")

print("\nОПЕРАЦИИ С ВЕКТОРАМИ И МАТРИЦАМИ")

# Сложение векторов
print("\n📈 Сложение векторов:")
вектор_A = np.array([1, 2, 3])
вектор_B = np.array([4, 5, 6])
сумма = вектор_A + вектор_B
print(f"{вектор_A} + {вектор_B} = {сумма}")

# Умножение вектора на число
print("\n📈 Умножение вектора на число:")
умноженный = вектор_A * 2
print(f"{вектор_A} * 2 = {умноженный}")

# Скалярное произведение (важно в машинном обучении!)
print("\n📈 Скалярное произведение (dot product):")
print("Это сумма произведений соответствующих элементов")
скалярное = np.dot(вектор_A, вектор_B)
print(f"{вектор_A} · {вектор_B} = {скалярное}")
print(f"Вычисление: 1*4 + 2*5 + 3*6 = {1*4 + 2*5 + 3*6}")

# Умножение матриц
print("\n📊 Умножение матриц (как в нейронных сетях!):")
матрица_A = np.array([[1, 2], [3, 4]])  # 2x2
матрица_B = np.array([[5, 6], [7, 8]])  # 2x2
произведение = np.dot(матрица_A, матрица_B)

print(f"Матрица A:\n{матрица_A}")
print(f"\nМатрица B:\n{матрица_B}")
print(f"\nПроизведение A × B:\n{произведение}")

print("\n🎮 ИНТЕРАКТИВНОЕ УМНОЖЕНИЕ МАТРИЦ")
print("Давай посчитаем, сколько денег нужно для покупок:")

# Создаем матрицу: количество товаров × цена
print("\nПредставим список покупок как матрицу:")
print("Строка 1: количество (яблоки, бананы, апельсины)")
print("Строка 2: цена за штуку (в рублях)")

количество = np.array([[3, 2, 4]])  # 3 яблока, 2 банана, 4 апельсина
цены = np.array([[30], [25], [40]])  # цены за штуку

print(f"\nКоличество: {количество}")
print(f"Цены:\n{цены}")

# Умножаем: количество × цены = общая стоимость
стоимость = np.dot(количество, цены)
print(f"\n💰 Общая стоимость: {стоимость[0][0]} руб.")
print(f"Расчет: 3*30 + 2*25 + 4*40 = {3*30 + 2*25 + 4*40} руб.")

print("\n" + "=" * 60)
```

```python
def создать_смайлик(эмоция='радость', размер_глаз=0.3, размер_рта=0.5):
    """
    Создает смайлик с заданными параметрами
    эмоция: радость, грусть, удивление, злость
    """
    fig, ax = plt.subplots(figsize=(6, 6))
    
    # Лицо (круг)
    круг = plt.Circle((0, 0), 1, color='yellow', ec='black', linewidth=2)
    ax.add_patch(круг)
    
    # Глаза (в зависимости от эмоции)
    if эмоция == 'удивление':
        # Большие круглые глаза
        левый_глаз = plt.Circle((-0.3, 0.3), размер_глаз, color='black')
        правый_глаз = plt.Circle((0.3, 0.3), размер_глаз, color='black')
    else:
        # Обычные глаза
        левый_глаз = plt.Circle((-0.3, 0.3), размер_глаз*0.7, color='black')
        правый_глаз = plt.Circle((0.3, 0.3), размер_глаз*0.7, color='black')
    
    ax.add_patch(левый_глаз)
    ax.add_patch(правый_глаз)
    
    # Рот (в зависимости от эмоции)
    x_рот = np.linspace(-0.7, 0.7, 100)
    
    if эмоция == 'радость':
        # Улыбка
        y_рот = -0.3 + размер_рта * 0.5 * (1 - (x_рот/0.7)**2)
        ax.plot(x_рот, y_рот, 'black', linewidth=3)
    elif эмоция == 'грусть':
        # Грустный рот
        y_рот = -0.6 - размер_рта * 0.3 * (1 - (x_рот/0.7)**2)
        ax.plot(x_рот, y_рот, 'black', linewidth=3)
    elif эмоция == 'удивление':
        # Круглый рот (удивление)
        рот = plt.Circle((0, -0.4), 0.2, color='black', fill=False, linewidth=3)
        ax.add_patch(рот)
    elif эмоция == 'злость':
        # Злой рот (перевернутая улыбка)
        y_рот = -0.6 + размер_рта * 0.3 * (1 - (x_рот/0.7)**2)
        ax.plot(x_рот, y_рот, 'black', linewidth=3)
        # Добавим брови
        ax.plot([-0.5, -0.1], [0.6, 0.5], 'black', linewidth=3)
        ax.plot([0.1, 0.5], [0.5, 0.6], 'black', linewidth=3)
    
    # Настройки отображения
    ax.set_xlim(-1.2, 1.2)
    ax.set_ylim(-1.2, 1.2)
    ax.set_aspect('equal')
    ax.axis('off')
    
    # Заголовок
    эмоции_русские = {'радость': '😊 Радость', 'грусть': '😢 Грусть', 
                     'удивление': '😮 Удивление', 'злость': '😠 Злость'}
    plt.title(f'Смайлик: {эмоции_русские.get(эмоция, эмоция)}', fontsize=14, pad=20)
    
    plt.show()
    print(f"✅ Создан смайлик с эмоцией: {эмоция}")

print("\n🎮 ПОИГРАЙ С ПАРАМЕТРАМИ СМАЙЛИКА!")
print("Измени эмоцию и размеры глаз/рта:")

# Интерактивные элементы
interact(создать_смайлик,
         эмоция=widgets.Dropdown(
             options=['радость', 'грусть', 'удивление', 'злость'],
             value='радость',
             description='Эмоция:',
             disabled=False
         ),
         размер_глаз=widgets.FloatSlider(
             value=0.3,
             min=0.1,
             max=0.5,
             step=0.05,
             description='Глаза:'
         ),
         размер_рта=widgets.FloatSlider(
             value=0.5,
             min=0.1,
             max=1.0,
             step=0.1,
             description='Рот:'
         ))

print("\n🎨 СОЗДАЙ СВОЙ УНИКАЛЬНЫЙ СМАЙЛИК!")
print("Попробуй разные комбинации и найди самую смешную!")

# Дополнительная функция для генерации случайного смайлика
import random

print("\n🎲 СЛУЧАЙНЫЙ СМАЙЛИК:")
случайная_эмоция = random.choice(['радость', 'грусть', 'удивление', 'злость'])
случайные_глаза = random.uniform(0.2, 0.4)
случайный_рот = random.uniform(0.3, 0.8)

print(f"Эмоция: {случайная_эмоция}, Глаза: {случайные_глаза:.2f}, Рот: {случайный_рот:.2f}")
создать_смайлик(случайная_эмоция, случайные_глаза, случайный_рот)

print("\n" + "=" * 60)
```

# 7. [Для интересующихся] Словари - как понятно и эффективно передавать значения?

```python
# Ячейка 8: 
print("\n📋 8. СЛОВАРИ - НАСТРОЙКИ ДЛЯ ФУНКЦИЙ")
print("-" * 40)

print("""В машинном обучении часто используют словари для хранения
настроек модели. Это как панель управления у робота!""")

# Пример: словарь настроек для робота-уборщика
print("\n🤖 Пример 1: Настройки робота-уборщика")

настройки_робота = {
    "name": "Уборщик-3000",
    "скоheightь": 5,           # от 1 до 10
    "мощность_пылесоса": 8,  # от 1 до 10
    "автономность_часов": 3,
    "умный_режим": True,
    "цвет": "синий",
    "задачи": ["пылесосить", "мыть пол", "вытирать пыль"]
}

print("Настройки робота:")
for ключ, значение in настройки_робота.items():
    print(f"  {ключ}: {значение}")

# Функция с конфигурацией через словарь
print("\n🔧 Пример 2: Функция рисования с настройками")

def нарисовать_дом(параметры):
    """
    Рисует дом по заданным параметрам
    """
    import matplotlib.pyplot as plt
    import matplotlib.patches as patches
    
    fig, ax = plt.subplots(figsize=(8, 6))
    
    # Основа дома (прямоугольник)
    основа = patches.Rectangle(
        (параметры["позиция_x"], параметры["позиция_y"]),
        параметры["ширина"], параметры["высота"],
        facecolor=параметры["цвет_стен"],
        edgecolor='black',
        linewidth=2
    )
    ax.add_patch(основа)
    
    # Крыша (треугольник)
    крыша = patches.Polygon(
        [[параметры["позиция_x"], параметры["позиция_y"] + параметры["высота"]],
         [параметры["позиция_x"] + параметры["ширина"], параметры["позиция_y"] + параметры["высота"]],
         [параметры["позиция_x"] + параметры["ширина"]/2, параметры["позиция_y"] + параметры["высота"] + параметры["высота_крыши"]]],
        facecolor=параметры["цвет_крыши"],
        edgecolor='black',
        linewidth=2
    )
    ax.add_patch(крыша)
    
    # Окно
    окно = patches.Rectangle(
        (параметры["позиция_x"] + параметры["ширина"]*0.7, 
         параметры["позиция_y"] + параметры["высота"]*0.5),
        параметры["размер_окна"], параметры["размер_окна"],
        facecolor=параметры["цвет_окна"],
        edgecolor='black',
        linewidth=2
    )
    ax.add_patch(окно)
    
    # Дверь
    дверь = patches.Rectangle(
        (параметры["позиция_x"] + параметры["ширина"]*0.2, 
         параметры["позиция_y"]),
        параметры["ширина_двери"], параметры["высота_двери"],
        facecolor=параметры["цвет_двери"],
        edgecolor='black',
        linewidth=2
    )
    ax.add_patch(дверь)
    
    # Настройки графика
    ax.set_xlim(0, 15)
    ax.set_ylim(0, 15)
    ax.set_aspect('equal')
    ax.axis('off')
    
    plt.title(f"Дом '{параметры.get('название', 'Мой дом')}'", fontsize=14, pad=20)
    plt.show()
    
    # Вычисляем площадь дома
    площадь = параметры["ширина"] * параметры["высота"]
    print(f"📐 Площадь дома: {площадь:.1f} кв.м")

print("\n🏠 СОЗДАЕМ ДОМ ПО НАШИМ ПАРАМЕТРАМ...")

# Параметры дома
мои_параметры = {
    "название": "Виллидж",
    "позиция_x": 5,
    "позиция_y": 3,
    "ширина": 6,
    "высота": 4,
    "высота_крыши": 3,
    "цвет_стен": "lightblue",
    "цвет_крыши": "red",
    "цвет_окна": "yellow",
    "цвет_двери": "brown",
    "размер_окна": 1,
    "ширина_двери": 1.2,
    "высота_двери": 2
}

нарисовать_дом(мои_параметры)

print("\n🎨 ИЗМЕНИ ПАРАМЕТРЫ ДОМА!")
print("Попробуй изменить значения в словаре 'мои_параметры' и перезапусти ячейку.")
print("Можно менять размеры, цвета, положение дома!")

print("\n" + "=" * 60)
```

# 8. [Для интересующихся] Вероятность - как часто встречаются значения?

Ключевые слова:
- вероятность
- частотность
- распределение

```python
import random
import matplotlib.pyplot as plt
```

```python
print("\n🪙 Пример 1: Подбрасывание монетки")
print("Теория говорит: вероятность орла = 0.5, решки = 0.5")
print("Давай проверим на практике!")

def подбросить_монетку(n_бросков):
    """Подбрасывает монетку n раз и считает результаты"""
    результаты = []
    орлы = 0
    решки = 0
    
    for i in range(n_бросков):
        if random.random() < 0.5:  # random() дает число от 0 до 1
            результаты.append("Орел")
            орлы += 1
        else:
            результаты.append("Решка")
            решки += 1
    
    # Вычисляем вероятности
    p_орлы = орлы / n_бросков
    p_решки = решки / n_бросков
    
    return результаты, p_орлы, p_решки, орлы, решки

print("\n🔹 Эксперимент 1: 10 бросков")
результаты_10, p_орлы_10, p_решки_10, орлы_10, решки_10 = подбросить_монетку(10)
print(f"Орлы: {орлы_10} ({p_орлы_10:.1%}), Решки: {решки_10} ({p_решки_10:.1%})")

print("\n🔹 Эксперимент 2: 100 бросков")
результаты_100, p_орлы_100, p_решки_100, орлы_100, решки_100 = подбросить_монетку(100)
print(f"Орлы: {орлы_100} ({p_орлы_100:.1%}), Решки: {решки_100} ({p_решки_100:.1%})")

print("\n🔹 Эксперимент 3: 1000 бросков")
результаты_1000, p_орлы_1000, p_решки_1000, орлы_1000, решки_1000 = подбросить_монетку(1000)
print(f"Орлы: {орлы_1000} ({p_орлы_1000:.1%}), Решки: {решки_1000} ({p_решки_1000:.1%})")

print("\n📊 ВИЗУАЛИЗАЦИЯ РЕЗУЛЬТАТОВ")
print("Чем больше бросков, тем ближе к 50%!")

# Строим график
броски = [10, 100, 1000]
процент_орлов = [p_орлы_10 * 100, p_орлы_100 * 100, p_орлы_1000 * 100]
процент_решек = [p_решки_10 * 100, p_решки_100 * 100, p_решки_1000 * 100]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# График 1: Проценты
x = range(len(броски))
ax1.plot(x, процент_орлов, 'bo-', linewidth=2, markersize=8, label='Орлы')
ax1.plot(x, процент_решек, 'ro-', linewidth=2, markersize=8, label='Решки')
ax1.axhline(y=50, color='green', linestyle='--', alpha=0.5, label='Теория (50%)')
ax1.set_xticks(x)
ax1.set_xticklabels(['10 бросков', '100 бросков', '1000 бросков'])
ax1.set_ylabel('Процент, %')
ax1.set_title('Частотность орлов и решек')
ax1.legend()
ax1.grid(True, alpha=0.3)

# График 2: Количество
ax2.bar(['10 бросков', '100 бросков', '1000 бросков'], 
        [орлы_10, орлы_100, орлы_1000], 
        color='blue', alpha=0.6, label='Орлы')
ax2.bar(['10 бросков', '100 бросков', '1000 бросков'], 
        [решки_10, решки_100, решки_1000], 
        bottom=[орлы_10, орлы_100, орлы_1000],
        color='red', alpha=0.6, label='Решки')
ax2.set_ylabel('Количество')
ax2.set_title('Абсолютное количество')
ax2.legend()

plt.tight_layout()
plt.show()

print("\n🎯 ЗАКЛЮЧЕНИЕ:")
print("Частотность ≈ Вероятность при большом количестве испытаний")
print("Это называется ЗАКОН БОЛЬШИХ ЧИСЕЛ")
print("\nВ машинном обучении это используется, чтобы понять:")
print("• Какие данные встречаются чаще")
print("• Какие ответы вероятнее")
print("• Как делать предсказания")

print("\n🎮 СИМУЛЯТОР КУБИКА")
print("Давай посмотрим, как часто выпадают разные числа на кубике!")

from ipywidgets import interact, widgets

@interact(n_бросков=widgets.IntSlider(value=100, min=10, max=1000, step=10, description='Бросков:'))
def симулятор_кубика(n_бросков=100):
    """Симулятор бросков кубика"""
    результаты = [random.randint(1, 6) for _ in range(n_бросков)]
    
    # Считаем частотность
    частотность = {i: 0 for i in range(1, 7)}
    for число in результаты:
        частотность[число] += 1
    
    # Вычисляем вероятности
    вероятности = {число: count/n_бросков for число, count in частотность.items()}
    
    # Выводим результаты
    print(f"\n🎲 Результаты {n_бросков} бросков кубика:")
    for число in range(1, 7):
        print(f"Число {число}: {частотность[число]} раз ({вероятности[число]:.1%})")
    
    # Теоретическая вероятность
    print(f"\n📊 Теоретическая вероятность каждого числа: {1/6:.1%}")
    
    # Визуализация
    fig, ax = plt.subplots(figsize=(10, 5))
    
    числа = list(range(1, 7))
    частоты = [частотность[i] for i in числа]
    теор_вероятность = [n_бросков/6 for _ in числа]  # Ожидаемое количество
    
    x = np.arange(len(числа))
    width = 0.35
    
    ax.bar(x - width/2, частоты, width, color='skyblue', label='Эксперимент')
    ax.bar(x + width/2, теор_вероятность, width, color='orange', alpha=0.5, label='Теория')
    
    ax.set_xlabel('Число на кубике')
    ax.set_ylabel('Количество выпадений')
    ax.set_title(f'Частотность выпадений кубика ({n_бросков} бросков)')
    ax.set_xticks(x)
    ax.set_xticklabels(числа)
    ax.legend()
    ax.grid(True, alpha=0.3)
    
    plt.show()

print("\n" + "=" * 60)
```
