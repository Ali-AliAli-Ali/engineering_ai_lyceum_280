# Создание игры "Виселица"

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
import random                             # для случайного выбора слова

def hangman():
    list_words=['программа',
                'питон',
                'наука',
                'образование',
                'школа',                    # список со словами
                'информатика',
                'математика']
    wrongs=0                               # кол-во ошибок
    stages=["",
            "________      ",
            "|      |      ",
            "|      0      ",              # стадии
            "|     /|\     ",
            "|     / \     ",
            "|"]

    word=random.choice(list_words)

    letters=list(word)                     # список угаданных букв, которые будут помечаться как $, пользователь не видит
    board=['__']*len(word)                 # доска угаданных букв, пользователь видит
    win=False                              # статус победы

    print('\n\nДобро пожаловать в игру!\n')
    print(' '.join(board))

    while wrongs<len(stages)-1:            # цикл, код выполняется когда кол-во ошибок меньше стадий

        msg=input('Введите букву (Пробел в слове тоже может быть!): ')      # ввод буквы
        if msg in letters:
          while msg in letters:                                                  # если мы угадываем букву
              index=letters.index(msg)
              board[index]=msg
              letters[index]='$'

        else:                                                              # если мы не угадываем букву
            wrongs+=1

        print(' '.join(board))                # вывод угаданной буквы и стадий
        e=wrongs+1
        print('\n'.join(stages[:e]))

        if '__' not in board:                                           # если мы угадали все буквы
            print('Вы выйграли! Было загадано слово: {}'.format(word))
            print(' '.join(board))
            win=True
            break

    if not win:
        print('\n'.join(stages))
        print('\nВы проиграли! Было загадано слово: {}'.format(word))  # пройгрыш

hangman()
```

# Объединим всё вместе

```python
# Список слов
all_words = [
    "мама",
    "папа",
    "плейстейшн",
    "макароны"
]

# Выбираем случайное слово
main_word=random.choice(all_words)

# Преобразовываем в список
letters=list(main_word)

# Создаём доску
board = ["_"] * len(main_word)

print("Добро пожаловать в игру 'Угадай слово'!")
print(" ".join(board))

wrong = 0

while (wrong < (len(stages) - 1)) and ("_" in board):
    # Просим пользователя ввести букву
    letter = str(input("Введите букву: ")).lower()

    if letter in letters:
        ind = letters.index(letter)
        board[ind] = letter
        letters[ind] = "$"
    else:
        wrong += 1

    # Отображаем текущее состояние поля
    print(" ".join(board))

    # Отображаем текущее состояние игрока
    print(stages[wrong])

if (wrong >= (len(stages) - 1)):
    print("Вы проиграли!")
else:
    print(f"Поздравляю с победой! Было загадано слово {main_word}")

```
